```python
# Load the corpus
from urllib.request import urlopen

url = "https://raw.githubusercontent.com/coli-saar/cl/main/a1/junglebook.txt"
text = urlopen(url).read().decode("utf-8")

# Remove metadata and noise, so that the model is trained on the narrative text only
start = text.find("Now Rann, the Kite")
end = text.find("Transcriber's Notes:")

text = text[start:end]

import re
text = re.sub(r"\[Illustration\]", "", text)
text = re.sub(r"\[Illustration:.*?\]", "", text, flags=re.DOTALL)

words = text.split()
```

Metadata and non-narrative content were removed from the corpus in order to train the language models only on the main text. 
Illustration markers were also removed during the preprocessing.
Unlike the previous analyses, the corpus was not converted to lowercase. Preserving capitalization allows the generated text to appear more natural and closer to the original text.

```python
# Import class for training
from ngram import BasicNgram
```

```python
# Train three models
bigram = BasicNgram(2, words)
trigram = BasicNgram(3, words)
fourgram = BasicNgram(4, words)
```

Three n-gram language models were trained on the corpus: a bigram model (n=2), a trigram model (n=3), and a four-gram model (n=4).

```python
# Create a text generator
def generate_text(model, n, length=100):

    context = ("<$>",) * (n - 1)  # creating the starting context

    generated = []

    for _ in range(length):

        prob_dist = model[context]  # what is next given the context

        next_word = prob_dist.generate()  # picks one word randomly according to the learned probabilities

        generated.append(next_word)

        context = context[1:] + (next_word,) # updating context

    return " ".join(generated)
```

A text generation function was implemented. Starting from the sentence-beginning context, the next word was repeatedly sampled according to the probability distribution learned by the n-gram model.

```python
# Bigram
text2 = generate_text(bigram, 2)
print(text2)

# Trigram
text3 = generate_text(trigram, 3)
print(text3)

# Fourgram
text4 = generate_text(fourgram, 4)
print(text4)
```

```python
# Generated text samples: 3 examples per model
for i in range(3):
    print("\nBIGRAM SAMPLE", i+1)
    print(generate_text(bigram, 2))

for i in range(3):
    print("\nTRIGRAM SAMPLE", i+1)
    print(generate_text(trigram, 3))

for i in range(3):
    print("\nFOURGRAM SAMPLE", i+1)
    print(generate_text(fourgram, 4))
```

Three n-gram models were trained on The Jungle Book corpus: a bigram model (n=2), a trigram model (n=3), and a four-gram model (n=4). 
For each model, several text samples of 100 words were generated.

The bigram model produced the most creative but also the least coherent outputs. Individual word transitions were often plausible, 
but the generated text frequently changed topic and lacked grammatical structure over longer sequences.

The trigram model generated more coherent text. Short phrases and fragments of sentences often resembled natural language, 
and local grammatical patterns were preserved more successfully than in the bigram model. At the same time, the text still contained 
unexpected transitions, which contributed to a certain degree of creativity.

The four-gram model produced the most fluent and coherent outputs. Longer sequences of words from the original corpus were reproduced, 
resulting in text that often appeared grammatical and locally meaningful. However, this improvement in quality came at the cost of 
reduced creativity, since the model relied more heavily on contexts observed in the training corpus.

Overall, increasing n improved coherence and grammaticality, while reducing randomness and novelty. 
This illustrates the trade-off between creativity and contextual consistency in n-gram language models.
